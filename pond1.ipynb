**importing the libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

**importing the dataset**

```python
data=pd.read_csv('IoTpond1.csv',parse_dates=['created_at'])
data.head()
```

```python
columns_to_drop=['entry_id','Population']
data.drop(columns=columns_to_drop,axis=1,inplace=True)
```

```python
data['date']=data['created_at'].dt.date
data['time']=data['created_at'].dt.time
data.head()
```

```python
columns=data.columns.to_list()
columns=columns[-2:] + columns[:-2]
data=data[columns]
data.drop('created_at',axis=1,inplace=True)
data.head()
```

```python
data.shape
```

```python
data.isna().sum()
```

```python
data['Ammonia(g/ml)'].mode()
```

```python
data['Ammonia(g/ml)'].replace([-np.inf,np.inf],np.nan,inplace=True)
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(missing_values=np.nan,strategy='median')
data[['Ammonia(g/ml)']]=imputer.fit_transform(data[['Ammonia(g/ml)']])
data[['Fish_Length(cm)']]=imputer.fit_transform(data[['Fish_Length(cm)']])
data[['Fish_Weight(g)']]=imputer.fit_transform(data[['Fish_Weight(g)']])
```

```python
data["Ammonia(g/ml)"] = data["Ammonia(g/ml)"].replace(
    [x for x in data["Ammonia(g/ml)"] if x<0 or x > 0.21],0.2)
```

```python
data["Nitrate(g/ml)"] = data["Nitrate(g/ml)"].replace(
    [x for x in data["Nitrate(g/ml)"] if  x >400 or x<0], data['Nitrate(g/ml)'].median())
```

```python
data["Turbidity(NTU)"] = data["Turbidity(NTU)"].replace(
    [x for x in data["Turbidity(NTU)"] if x > 100 or x < 30], data['Turbidity(NTU)'].median())
```

```python
data["Temperature (C)"] = data["Temperature (C)"].replace(
    [x for x in data["Temperature (C)"] if x > 30 or x < 21], data['Temperature (C)'].median())
```

```python
data["PH"] = data["PH"].replace(
    [x for x in data["PH"] if x > 9 or x < 6.5], data['PH'].median())
```

```python
data["Dissolved Oxygen(g/ml)"] = data["Dissolved Oxygen(g/ml)"].replace(
    [x for x in data["Dissolved Oxygen(g/ml)"] if x>20 or x < 5], data['Dissolved Oxygen(g/ml)'].median())
```

```python
data['date']=data['date'].astype('category')
data['date']=data['date'].cat.codes
data.drop('time',axis=1,inplace=True)
```

```python
x=data.iloc[:,:-2].values
y=data.iloc[:,-1].values
y=y.reshape(len(y),1)
```

```python
print(x)
```

```python
print(y)
```

```python
data.head()
```

splitting the dataset into training and testing

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
print(x_train)
```

training the model

```python
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(n_estimators=100,random_state=0)
regressor.fit(x_train,y_train)
```

predicting the model

```python
y_pred=regressor.predict(x_test)
```

```python
print(y_pred)
```

```python
print(y_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error
print(r2_score(y_test,y_pred))
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
```

```python
print('MSE=',mse)
print('RMSE=',rmse)
```
